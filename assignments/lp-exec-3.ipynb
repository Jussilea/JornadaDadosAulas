# Problem 1

The Petrisul oil refinery uses three types of crude oil (oil 1, oil 2, and oil 3) to produce three types of gasoline: regular, super, and extra.

To guarantee quality, each type of gasoline requires a composition of the various types of crude oil, as shown in the table below.

|Types of gasoline|Specs|
|--|--|
|Regular|No more than 70% of the oil 1|
|Super|No more than 50% of the oil 1|
||No less than 10% of the oil 2|
|Extra|No more than 50% of the oil 2|
||No less than 40% of the oil 3|

To meet customer demand, the refinery needs to produce at least 5,000 barrels per day of regular gasoline and 3,000 barrels per day of super and extra gasoline. The available daily capacity is 10,000 barrels of crude oil 1, 8,000 of crude oil 2, and 7,000 of crude oil 3. The refinery can produce up to 20,000 barrels of gasoline per day.

The refinery profits \$5 for each barrel of regular gasoline produced, \$7 per barrel of super gasoline, and $8 per barrel of extra gasoline. The costs of crude oil 1, 2, and 3 per barrel are \$2, \$3, and \$3, respectively. Formulate and solve the linear programming problem to maximize daily profit.

````python
!pip install -q pyomo
````

````python
from pyomo.environ import *

# create a model
model = ConcreteModel()

# declare decision variables - barris de petróleo
model.R_oil1 = Var(domain=NonNegativeReals)
model.R_oil2 = Var(domain=NonNegativeReals)
model.R_oil3 = Var(domain=NonNegativeReals)
model.S_oil1 = Var(domain=NonNegativeReals)
model.S_oil2 = Var(domain=NonNegativeReals)
model.S_oil3 = Var(domain=NonNegativeReals)
model.E_oil1 = Var(domain=NonNegativeReals)
model.E_oil2 = Var(domain=NonNegativeReals)
model.E_oil3 = Var(domain=NonNegativeReals)

# declare objective
model.lucro = Objective(expr = 5*(model.R_oil1 + model.R_oil2 + model.R_oil3) + 7*(model.S_oil1 + model.S_oil2 + model.S_oil3) + 8*(model.E_oil1 + model.E_oil2 + model.E_oil3)
- 2*(model.R_oil1 + model.S_oil1 + model.E_oil1) - 3*(model.R_oil2 + model.S_oil2 + model.E_oil2) - 3*(model.R_oil3 + model.S_oil3 + model.E_oil3), sense=maximize)

# declare constraints
model.regular_comp = Constraint(expr = model.R_oil1 <= 0.7*(model.R_oil1 + model.R_oil2 + model.R_oil3))
model.super_comp_oil1 = Constraint(expr = model.S_oil1 <= 0.5*(model.S_oil1 + model.S_oil2 + model.S_oil3))
model.super_comp_oil2 = Constraint(expr = model.S_oil2 >= 0.1*(model.S_oil1 + model.S_oil2 + model.S_oil3))
model.extra_comp_oil2 = Constraint(expr = model.E_oil2 <= 0.5*(model.E_oil1 + model.E_oil2 + model.E_oil3))
model.extra_comp_oil3 = Constraint(expr = model.E_oil3 >= 0.4*(model.E_oil1 + model.E_oil2 + model.E_oil3))
model.demanda_regular = Constraint(expr = model.R_oil1 + model.R_oil2 + model.R_oil3 >= 5000)
model.demanda_super = Constraint(expr = model.S_oil1 + model.S_oil2 + model.S_oil3 >= 3000)
model.demanda_extra = Constraint(expr = model.E_oil1 + model.E_oil2 + model.E_oil3 >= 3000)
model.disp_oil1 = Constraint(expr = model.R_oil1 + model.S_oil1 + model.E_oil1 <= 10000)
model.disp_oil2 = Constraint(expr = model.R_oil2 + model.S_oil2 + model.E_oil2 <= 8000)
model.disp_oil3 = Constraint(expr = model.R_oil3 + model.S_oil3 + model.E_oil3 <= 7000)
model.capacidade = Constraint(expr = (model.R_oil1 + model.R_oil2 + model.R_oil3) + (model.S_oil1 + model.S_oil2 + model.S_oil3) + (model.E_oil1 + model.E_oil2 + model.E_oil3) <= 20000)

model.pprint()
````

````python
!apt install coinor-cbc
````

````python
SolverFactory('cbc', executable='/usr/bin/cbc').solve(model).write()

# display solution
print('\nLucro = ', model.lucro())

print('\nDecision Variables')
print('R_oil1 = ', model.R_oil1())
print('R_oil2 = ', model.R_oil2())
print('R_oil3 = ', model.R_oil3())
print('S_oil1 = ', model.S_oil1())
print('S_oil2 = ', model.S_oil2())
print('S_oil3 = ', model.S_oil3())
print('E_oil1 = ', model.E_oil1())
print('E_oil2 = ', model.E_oil2())
print('E_oil3 = ', model.E_oil3())

print('\nConstraints')
print('Composição Regular  = ', model.regular_comp())
print('Composição Super oil1 = ', model.super_comp_oil1())
print('Composição Super oil2 = ', model.super_comp_oil2())
print('Composição Extra oil2 = ',model.extra_comp_oil2())
print('Composição Extra oil3 = ',model.extra_comp_oil3())
print('Demanda Regular = ',model.demanda_regular())
print('Demanda Super = ',model.demanda_super())
print('Demanda Extra = ',model.demanda_extra())
print('Disponibilidade oil1 = ',model.disp_oil1())
print('Disponibilidade oil2 = ',model.disp_oil2())
print('Disponibilidade oil3 = ',model.disp_oil3())
print('Capacidade total = ',model.capacidade())
````

# Problem 2

CPS Investment Services must develop an investment portfolio for a new customer. Initially, the new customer would like to restrict portfolio to three stocks as shown below:

|Stock|Price per Share ($)|Estimated Annual Return (%)|
|--|--|--|
|ACT cable|50|6|
|CTR retail|35|10|
|PCL petroleum|120|8|

The customer wants to invest $80,000 and established two investment
goals.
Goal 1: Obtain an annual return of at least 9%.
Goal 2: Limit the investment in CTR retail, the riskier investment, to no more than 50% of the total investment.

Formulate a mathematical programming model for the CPS investment problem and solve it.

````python
from pyomo.environ import *

# create a model
model = ConcreteModel()

# declare decision variables
model.act = Var(domain=NonNegativeReals)
model.ctr = Var(domain=NonNegativeReals)
model.pcl = Var(domain=NonNegativeReals)

# declare objective
model.lucro = Objective(expr = 0.06*(50*model.act) + 0.10*(35*model.ctr) + 0.08*(120*model.pcl), sense=maximize)

# declare constraints
model.valor_aplicacao = Constraint(expr = 50*model.act + 35*model.ctr + 120*model.pcl <= 80000)
model.retorno_anual = Constraint(expr = 0.06*(50*model.act) + 0.10*(35*model.ctr) + 0.08*(120*model.pcl) >= 0.09*(50*model.act + 35*model.ctr + 120*model.pcl))
model.aplicação_ctr = Constraint(expr = 35*model.ctr <= 0.5*(50*model.act + 35*model.ctr + 120*model.pcl))

model.pprint()
````

````python
SolverFactory('cbc', executable='/usr/bin/cbc').solve(model).write()

# display solution
print('\nLucro = ', model.lucro())

print('\nDecision Variables')
print('ACT = ', model.act())
print('ctr = ', model.ctr())
print('pcl = ', model.pcl())

print('\nConstraints')
print('Valor aplicado  = ', model.valor_aplicacao())
print('Retorno anual = ', model.retorno_anual())
print('Aplicação CTR = ', model.aplicação_ctr())
````

# Problem 3

A popular product is produced at three plants (PL1, PL2, PL3) and is required to ship to three warehouses: WH1, WH2, and WH3. The supply from each plant, the delivery required in each warehouse, and the unit transportation cost from each plant to each warehouse is shown below:

|Plant|WH1|WH2|WH3|Plant Capacity|
|--|--|--|--|--|
|PL1|20|15|25|400|
|PL2|10|12|8|500|
|PL3|14|20|12|200|
|Warehouse demand|300|200|600||

Develop an LP model for minimizing the transportation cost while not violating the supply and demand constraints.

````python
from pyomo.environ import *

# create a model
model = ConcreteModel()

# declare decision variables
model.PL1WH1 = Var(domain=NonNegativeReals)
model.PL1WH2 = Var(domain=NonNegativeReals)
model.PL1WH3 = Var(domain=NonNegativeReals)
model.PL2WH1 = Var(domain=NonNegativeReals)
model.PL2WH2 = Var(domain=NonNegativeReals)
model.PL2WH3 = Var(domain=NonNegativeReals)
model.PL3WH1 = Var(domain=NonNegativeReals)
model.PL3WH2 = Var(domain=NonNegativeReals)
model.PL3WH3 = Var(domain=NonNegativeReals)

# declare objective
model.custo_transp = Objective(expr = 20*model.PL1WH1 + 15*model.PL1WH2 + 25*model.PL1WH3 + 10*model.PL2WH1 + 12*model.PL2WH2 + 8*model.PL2WH3 + 14*model.PL3WH1 + 20*model.PL3WH2 + 12*model.PL3WH3 , sense=minimize)

# declare constraints
model.capacidadePL1 = Constraint(expr = model.PL1WH1 + model.PL1WH2 + model.PL1WH3 <= 400)
model.capacidadePL2 = Constraint(expr = model.PL2WH1 + model.PL2WH2 + model.PL2WH3 <= 500)
model.capacidadePL3 = Constraint(expr = model.PL3WH1 + model.PL3WH2 + model.PL3WH3 <= 200)
model.demandaWH1 = Constraint(expr = model.PL1WH1 + model.PL2WH1 + model.PL3WH1 == 300)
model.demandaWH2 = Constraint(expr = model.PL1WH2 + model.PL2WH2 + model.PL3WH2 == 200)
model.demandaWH3 = Constraint(expr = model.PL1WH3 + model.PL2WH3 + model.PL3WH3 == 600)

model.pprint()
````

````python
SolverFactory('cbc', executable='/usr/bin/cbc').solve(model).write()

# display solution
print('\nCusto do Transporte = ', model.custo_transp())

print('\nDecision Variables')
print('PL1WH1 = ', model.PL1WH1())
print('PL1WH2 = ', model.PL1WH2())
print('PL1WH3 = ', model.PL1WH3())
print('PL2WH1 = ', model.PL2WH1())
print('PL2WH2 = ', model.PL2WH2())
print('PL2WH3 = ', model.PL2WH3())
print('PL3WH1 = ', model.PL3WH1())
print('PL3WH2 = ', model.PL3WH2())
print('PL3WH3 = ', model.PL3WH3())

print('\nConstraints')
print('CapacidadePL1  = ', model.capacidadePL1())
print('CapacidadePL2  = ', model.capacidadePL2())
print('CapacidadePL3  = ', model.capacidadePL3())
print('DemandaWH1  = ', model.demandaWH1())
print('DemandaWH2  = ', model.demandaWH2())
print('DemandaWH3  = ', model.demandaWH3())
````

# Problem 4

A certain paint can be produced using four different production processes (P1, P2, P3, and P4). The processing cost of each liter in any of the four available processes, the maximum capacity of each process, and the setup costs are given below:

|Process|Setup Cost (\$)|Processing Cost (\$/L)|Capacity (L)|
|--|--|--|--|
|P1|5,000|0.60|20,000|
|P2|6,000|0.50|15,000|
|P3|10,000|0.40|40,000|
|P4|6,000|0.40|25,000|

Assume that a daily demand of 45,000 L must be fulfilled. Formulate the problem as a programming model to determine the daily production schedule in order to minimize total costs.

````python
from pyomo.environ import *

# create a model
model = ConcreteModel()

# declare decision variables
model.P1 = Var(domain=NonNegativeReals)
model.P2 = Var(domain=NonNegativeReals)
model.P3 = Var(domain=NonNegativeReals)
model.P4 = Var(domain=NonNegativeReals)

# declaração variáveis de ativação do processo
model.P1S = Var(domain=Binary)
model.P2S = Var(domain=Binary)
model.P3S = Var(domain=Binary)
model.P4S = Var(domain=Binary)

# declare objective
model.custo = Objective(expr = 0.60*model.P1 + 0.50*model.P2 + 0.40*model.P3 + 0.40*model.P4 + 5000*model.P1S + 6000*model.P2S + 10000*model.P3S + 6000*model.P4S , sense=minimize)

# declare constraints
model.demanda = Constraint(expr = model.P1 + model.P2 + model.P3 + model.P4 == 45000)
model.capacidadeP1 = Constraint(expr = model.P1 <= 20000*model.P1S)
model.capacidadeP2 = Constraint(expr = model.P2 <= 15000*model.P2S)
model.capacidadeP3 = Constraint(expr = model.P3 <= 40000*model.P3S)
model.capacidadeP4 = Constraint(expr = model.P4 <= 25000*model.P4S)


model.pprint()
````

````python
SolverFactory('cbc', executable='/usr/bin/cbc').solve(model).write()

# display solution
print('\nCusto Total = ', model.custo())

print('\nDecision Variables')
print('P1 = ', model.P1())
print('P2 = ', model.P2())
print('P3 = ', model.P3())
print('P4 = ', model.P4())


print('\nConstraints')
print('Demanda  = ', model.demanda())
print('CapacidadeP1  = ', model.capacidadeP1())
print('CapacidadeP2  = ', model.capacidadeP2())
print('CapacidadeP3  = ', model.capacidadeP2())
print('CapacidadeP4  = ', model.capacidadeP2())
````
